```python
class Task:
    def __init__(self, title, priority):
        self.title = title
        self.priority = priority
        self.is_done = False


class TodoList:
    def __init__(self):
        self.tasks = []

    def add_task(self, task):
        self.tasks.append(task)

    def get_tasks(self):
        return self.tasks


def main():
    todo = TodoList()

    title = input("Enter task title: ").strip()

    if not title:
        print("Error: Task title cannot be empty.")
        return

    priority = input("Enter task priority: ").strip()

    task = Task(title, priority)
    todo.add_task(task)

    print("\nYour tasks:")

    for task in todo.get_tasks():
        print(f"- {task.title} | Priority: {task.priority}")


main()
```

```python
def get_column_average(filename, col_index):
    try:
        with open(filename, 'r') as f:
            data = f.readlines()
            total = 0
            count = 0
            for line in data[1:]:
                values = line.split(',')
                total += float(values[col_index])
                count += 1
            return total / count
    except IOError:
        print('File not found')
    except ValueError:
        print('Invalid data')

# Usage example:
# print(get_column_average('data.csv', 1))
```

```python
import csv

def get_column_average(filename, col_index):
    try:
        with open(filename, "r") as file:
            reader = csv.reader(file)

            next(reader)  # Skip the header

            values = []

            for row in reader:
                try:
                    values.append(float(row[col_index]))
                except ValueError:
                    print("Invalid data")
                    return None

            return sum(values) / len(values)

    except IOError:
        print("File not found")
        return None
```
