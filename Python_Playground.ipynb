```python
#THIS is a seperate project
#from a self learning app, called codino
#Fact, I just copied the solution


#Define decorator and wrapper
def log_activity(func):
    def wrapper(*args, **kwargs):
        print("Starting operation...")
        return func(*args, **kwargs)
    return wrapper


#call the decorator with @
#Define your fnc and
@log_activity
def process_data(data_list):
    for item in data_list:
        yield item.upper()

data = ["apple", "banana", "cherry"]

#Call the fnc
for item in process_data(data):
    print(item)
```

```python
class Task:
    def __init__(self, title, priority):
        self.title = title
        self.priority = priority
        self.is_done = False


class TodoList:
    def __init__(self):
        self.tasks = []

    def add_task(self, task):
        self.tasks.append(task)

    def get_tasks(self):
        return self.tasks


def main():
    todo = TodoList()

    title = input("Enter task title: ").strip()

    if not title:
        print("Error: Task title cannot be empty.")
        return

    priority = input("Enter task priority: ").strip()

    task = Task(title, priority)
    todo.add_task(task)

    print("\nYour tasks:")

    for task in todo.get_tasks():
        print(f"- {task.title} | Priority: {task.priority}")


main()
```

```python
def get_column_average(filename, col_index):
    try:
        with open(filename, 'r') as f:
            data = f.readlines()
            total = 0
            count = 0
            for line in data[1:]:
                values = line.split(',')
                total += float(values[col_index])
                count += 1
            return total / count
    except IOError:
        print('File not found')
    except ValueError:
        print('Invalid data')

# Usage example:
# print(get_column_average('data.csv', 1))
```

```python
import csv

def get_column_average(filename, col_index):
    try:
        with open(filename, "r") as file:
            reader = csv.reader(file)

            next(reader)  # Skip the header

            values = []

            for row in reader:
                try:
                    values.append(float(row[col_index]))
                except ValueError:
                    print("Invalid data")
                    return None

            return sum(values) / len(values)

    except IOError:
        print("File not found")
        return None
```

Say you are helping a friend to calculate their weekly paycheck after taxes.

They're in a 12% tax bracket (in other words, 12% of their salary is taken for taxes, and they only take home 88%), and
They're paid hourly, at a rate of $15/hour.
The function below calculates the paycheck based on the number of hours worked. The function is more complicated than with the first example, because the function has more lines of code and comments. Similar to the example above, the function has a single argument (num hours). In the function body, we:

Use the value for num_hours to specify the value for a new variable pay_pretax.
Use the value of pay_pretax to specify the value for a new variable pay_aftertax.
Return the value of the pay_aftertax variable.



```python
def get_pay(num_hrs): #define function to calculate pretax pay from total ours work
  pay_pretax = num_hrs * 15  # multiply the total hours with pay per hour $15
  pay_aftertax = pay_pretax * (1 - 0.12) # this line calculate the aftertax of .12
  return pay_aftertax # return the after tax

print(get_pay(50))
print(get_pay(40))

def main(): # define main function
   hrs_work = input("Enter hours work: ").strip()
   try:
      hrs_work = float(hrs_work)
      pay = get_pay(hrs_work)
      print(pay)
   except ValueError:
      print("Error, please enter numeric input")
main()
```

```python
def get_pay1(num_hrs, hrs_pay, tax):
  pay_pretax = num_hrs * hrs_pay
  pay_aftertax = pay_pretax * ( 1 - tax/100)
  return pay_aftertax

print(f"${get_pay1(40, 24, 22):.2f}")

def main():
  try:
    hrs_work = float(input("Enter hours work: ").strip())
    hrs_pay = float(input("Enter yuor pay per hour: ").strip())
    tax = float(input("Enter your tax percent i.e 10, 15, 20: ").strip())
    gross = get_pay1(hrs_work, hrs_pay, 0)
    pay = get_pay1(hrs_work, hrs_pay, tax)
    print(f" Gross pay: ${gross:.2f} \n Net pay after tax: ${pay:.2f}")
  except ValueError:
    print("Error, please enter numeric input.")
main()
```
