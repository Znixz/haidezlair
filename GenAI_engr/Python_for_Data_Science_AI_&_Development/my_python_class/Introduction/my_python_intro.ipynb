**Python Data Types**
As an obj oriented lang, common objects types include
str (strings) : the seguence of char in quotes i.e "this is a string"
float : float are essentially real numbers and are differentiate by the presence of decimal point i.e 5.0, 6.0, 1.43, 5.55.
int (integers) : are the set of number without decimal, say whole numbers from positive to negative i.e 1, -1, 2, -2, 7, -7

other form of data type include boolean which can take only two values that is , True and False. Not the first letter is capitalise.

```python
## Tell python to tell you your data type by using python type() command

print(type(9))
print(type(9.9))
print(type("9"))
print(type(True))
print(type(False))
```

```python
# both float and int has limit but int is quite large while float is quite small
# the code below check that though unnecessary

print(sys.int_info)
print(sys.float_info)
```

**Type casting** allows python to convert one obj type to another.
However string conversion is limited to integer/float string obj i.e "8787" , "445.334". And boolean conversion only rturns false for 0 or empty string


```python
print(int(4.6))
print(int(False))
print(int("2"))
print(int("3.55")) # error
print(int("hello")) ## this will return err
```

```python
print(float("1")) # str to float
print(float("1.2")) # str to float
print(float(2)) # int to float
print(float(True)) # bool to float
print(float(False))

## conversion to str
print(str(1))
print(str(1.2))
print(str(True))
print(str(False))


## conversion to bool
print(bool(1))
print(bool(0))
print(bool(1.2))
print(bool(""))
```
