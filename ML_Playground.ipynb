```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ucimlrepo import fetch_ucirepo
#fetch dataset
auto_mpg = fetch_ucirepo(id = 9)
# data as pandas dataframe
x = auto_mpg.data.features
y = auto_mpg.data.targets
z = auto_mpg.data.variables
xyz = auto_mpg.metadata


print(f"Features:\n{x[["weight"]].head(10)}\n\nTarget:\n{y[["mpg"]].head(10)}")

X = x[["weight"]]/1000
Y = y[["mpg"]]
table_XY = pd.concat([X, Y], axis = 1)

display(table_XY.head(10))

plt.scatter(X.head(10), Y.head(10))
plt.xlabel("Pounds in 000")
plt.ylabel("Mile per galon")
plt.show()
```

```python
sample = table_XY.head(77)
display(sample)

plt.scatter(sample["weight"], sample["mpg"])
plt.xlim(0,6)
plt.ylim(0,20)
plt.xlabel("Pound in 000 (Features)")
plt.ylabel("Miles per gallon (Label)")
plt.show()
```
